```python
from build123d import *
from ocp_vscode import *
from math import sin, cos,tan, pi
set_port(3939)

set_defaults(reset_camera=Camera.CENTER, helper_scale=5)
```

```python
length = 200
width = 300
thickness = 10.0
radius = length * tan(pi/6)

with BuildSketch() as profile_sk:
    with BuildLine():
        Line((length, 0), (length, width))
        Line((0, 0), (0, width))
        CenterArc(
            center = (length/2, width-radius/2),
            radius = radius,
            start_angle=30,
            arc_size=120,
        )
        CenterArc(
            center = (length/2, radius/2),
            radius = radius,
            start_angle=180+30,
            arc_size=120,
        )
    make_face()
show(profile_sk)
```
