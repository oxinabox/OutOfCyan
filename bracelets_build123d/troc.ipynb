```python
from build123d import *
from ocp_vscode import *
from math import sin, cos,tan, pi
set_port(3939)

set_defaults(reset_camera=Camera.CENTER, helper_scale=5)
```

```python
band_width = 50
end_cap_radius = band_width * tan(pi/6)

wrist_short_rad = 31
wrist_wide_rad = 37
band_thickness = 5

cap_angle = 30
# prep the shape for the cap
cap_arc=CenterArc(
    center = (end_cap_radius, 0),
    radius = end_cap_radius,
    start_angle=90 +cap_angle,
    arc_size=180 - 2*cap_angle,
    mode=Mode.PRIVATE
)
cap_butt = Line(cap_arc.start_point(), cap_arc.end_point(), mode=Mode.PRIVATE)
cap_sk = make_face([cap_arc, cap_butt], mode=Mode.PRIVATE)

# place things
with BuildPart() as p:
    with BuildLine() as l:
        wrist_arc = EllipticalCenterArc(
            center=(0,0),
            x_radius=wrist_wide_rad, y_radius=wrist_short_rad,
            start_angle=-70
        )
    with BuildSketch(l.wire() ^ 0):
        Ellipse(y_radius=band_width/2, x_radius=band_thickness/2)
    a = sweep()

    
    end_loc = wrist_arc @ 0.0 #front_face.location_at(0.4, 0)
    cap_normal = (wrist_arc % 0.0).rotate(Axis.Z, 90)
    cap_workplane = Plane(
        origin=end_loc, x_dir=(1, 0, 0), z_dir=cap_normal
    )
    
    anti_cap_sk = cap_workplane * (bounding_box(cap_sk) - cap_sk)
    extrude(anti_cap_sk, amount=-10, both=True, mode=Mode.SUBTRACT)
  
    mirror(about=Plane.YZ)


    # Project is mostly broken
    #with BuildSketch(Plane((0, 10+wrist_short_rad,0), x_dir=(1, 0, 0), z_dir=(0, 1, 0))):
    #    Text("CUNT", font_size=25, align=(Align.CENTER, Align.CENTER))
    #extrude(amount=-30, mode=Mode.SUBTRACT)
show(p)
```
