```python
from build123d import *
from ocp_vscode import *
from math import sin, cos,tan, pi
set_port(3939)

set_defaults(reset_camera=Camera.CENTER, helper_scale=5)
```

```python
from gridfinity_build123d import (
    BaseEqual,
    Bin,
    Compartment,
    CompartmentsEqual,
)
wall_thick=1
```

```python
draw_width = 510
draw_depth = 390
draw_height = 38
brush_length = 190
brush_compartment_width = 105
tall_misc_length = 190
tall_misc_width = 40
misc_length = 125
mirror_length = 155
mirrow_width = 140

gf = 42
gf_h =7
```

```python
def pen_bin(bin_width=1):
    with BuildPart() as part:
        Bin(
            BaseEqual(grid_x=bin_width, grid_y=4),
            height_in_units=6,
            compartments=CompartmentsEqual(compartment_list=[Compartment()]),
        )
        faces_by_x = part.faces().group_by(Axis.X)
        side_face, = faces_by_x[1]
        with BuildSketch(Plane.YZ):
            with BuildLine(Location((0,2.5))):
                arc = EllipticalCenterArc((0,0),  2*gf_h, gf, -90, rotation=90)
                Line(arc @ 0, arc @ 1)
            make_face()
        bump=extrude(amount=side_face.position.X)
        mirror(bump, Plane.YZ)
    return part
pen1 = pen_bin(bin_width=1)
show(pen1)
```

```python
pen2 = pen_bin(bin_width=2)
show(pen2)
```

```python
export_stl(pen1.part, "pen1.stl")
export_stl(pen2.part, "pen2.stl")
```
