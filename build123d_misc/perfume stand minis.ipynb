```python
from build123d import *
from ocp_vscode import *
from math import sin, cos,tan, pi
set_port(3939)
set_defaults(reset_camera=Camera.CENTER, helper_scale=5)
import numpy as np
```

```python
base_len = 80
base_wid = 45
hole_deep = 12
base_tall = hole_deep + 2


space = 0.2
r1 = 11.2/2 + space
r2 = 10.2/2 + space
r3 = 11.6/2 + space

with BuildPart() as part:
    with BuildSketch() as sk:
        RectangleRounded(base_len, base_wid, base_wid/4)

    extrude(amount=base_tall)
    chamfer(part.edges().group_by(Axis.Z)[-1], length=3)

    top_face = part.faces().sort_by(Axis.Z)[-1]
    top = top_face.location
    with BuildSketch(top*Location((0, 10, 0))):
        with GridLocations(17, 1, 4, 1):
            Circle(r1)
    with BuildSketch(top*Location((-16, -10, 0))):
        Circle(r2)
    with BuildSketch(top*Location((16, -10, 0))):
        Circle(r3)
    extrude(amount=-hole_deep, mode=Mode.SUBTRACT)

    top_edges = part.edges().group_by(Axis.Z)[-1]
    hole_edges = set(top_edges).difference(top_face.edges())
    fillet(hole_edges, 2)
    
show(part)
```

```python
export_stl(part.part, "perfume_mini_stand.stl")
```
