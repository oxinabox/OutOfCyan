```python
from build123d import *
from ocp_vscode import *
from math import sin, cos,tan, pi
set_port(3940)

set_defaults(reset_camera=Camera.CENTER, helper_scale=5)
```

```python
card_width = 91
card_height = 55
thick=2.4

with BuildPart() as p:
    with BuildSketch():
        RectangleRounded(card_width, card_height, 10)
        with Locations(((0, card_height/2 - 1.5))):
            Ellipse(7, 2, mode=Mode.SUBTRACT, align=(Align.CENTER, Align.MAX))
    extrude(amount=thick)
    with BuildSketch(Location((0, card_height/2 - 13, thick/2))):
        Text(
            "I AM\nTOTALLY\nDEAF IN MY\nRIGHT EAR",
            font_size=6.7, font_style=FontStyle.BOLD,
            #font_path="gunplay rg.otf", font_size=7.5, 
            align=(Align.MIN, Align.MAX),
        )
    extrude(amount=thick/2, mode=Mode.SUBTRACT)

    for sym_face in import_svg("symbol.svg")[1:]:
        face = scale(sym_face, 0.1).translate([-40, +15, thick/2])
        extrude(face, amount=thick/2, mode=Mode.SUBTRACT)
    
show(p)
```

```python
export_stl(p.part, "card.stl")
```
